```python
import matplotlib.pyplot as plt

# Sample Data
hours_studied = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [35, 40, 50, 55, 65, 70, 78, 85, 92]
subjects = ["Math", "Physics", "Chemistry", "Biology", "English"]
subject_scores = [85, 72, 90, 68, 80]

# Create a figure with 3 subplots
plt.figure(figsize=(15, 4))

# ---------------------------------------------------------
# 1. Line Chart: Study Hours vs Exam Scores Trend
# ---------------------------------------------------------
plt.subplot(1, 3, 1)
plt.plot(
    hours_studied,
    exam_scores,
    marker="o",
    color="b",
    linestyle="-",
    linewidth=2,
)
plt.title("Line Chart: Performance Trend")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.grid(True)

# ---------------------------------------------------------
# 2. Scatter Plot: Hours Studied vs Scores
# ---------------------------------------------------------
plt.subplot(1, 3, 2)
plt.scatter(hours_studied, exam_scores, color="red", s=50)
plt.title("Scatter Plot: Hours vs Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.grid(True)

# ---------------------------------------------------------
# 3. Bar Chart: Marks by Subject
# ---------------------------------------------------------
plt.subplot(1, 3, 3)
plt.bar(subjects, subject_scores, color="teal")
plt.title("Bar Chart: Subject Scores")
plt.xlabel("Subjects")
plt.ylabel("Score")
plt.grid(axis="y")

# Display the layout neatly
plt.tight_layout()
plt.show()
```
