```python
# ============================================
# DAY 9: PREDICTION TASKS
# ============================================

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# 1. Load the clean dataset
df = pd.read_csv("student_scores_clean.csv")

# 2. Separate Features (X) and Target (Y)
X = df[['Hours']]
Y = df['Scores']

# 3. Split Data & Train Model
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, Y_train)

# 4. Generate Predictions on the Test Set
Y_pred = model.predict(X_test)

# Display Actual vs Predicted comparison
comparison_df = pd.DataFrame({'Actual Score': Y_test, 'Predicted Score': Y_pred})
print("--- 1. Test Set Predictions Comparison ---")
print(comparison_df.to_string(index=False))

# 5. Predict for Custom Inputs (e.g., 9.25 hours & 4.5 hours)
custom_hours = [[9.25], [4.5]]
custom_preds = model.predict(custom_hours)

print("\n--- 2. Custom Predictions ---")
for hours, pred in zip(custom_hours, custom_preds):
    print(f"Studying for {hours[0]} hours -> Predicted Score: {pred:.2f}")
```
