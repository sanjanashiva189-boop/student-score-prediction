```python
# ============================================
# DAY 8: BUILD THE MODEL TASKS
# ============================================

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# 1. Load the clean dataset
df = pd.read_csv("student_scores_clean.csv")

# 2. Separate Features (X) and Target (Y)
X = df[['Hours']]
Y = df['Scores']

# 3. Train-Test Split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

# 4. Instantiate and Train the Linear Regression Model
model = LinearRegression()
model.fit(X_train, Y_train)

print("--- Day 8: Model Training Complete ---")
print(f"Model Intercept (c): {model.intercept_:.2f}")
print(f"Model Coefficient/Slope (m): {model.coef_[0]:.2f}")
print("\nModel Formula: Predicted Score =", f"{model.coef_[0]:.2f} * Hours + {model.intercept_:.2f}")
```
