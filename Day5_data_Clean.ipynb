```python
# ============================================
# DAY 5: DATA CLEANING TASKS
# ============================================

import pandas as pd

# 1. Load Dataset
df = pd.read_csv("student_scores.csv")

print("--- 1. Initial Dataset ---")
print(df)
print("\nInitial Shape:", df.shape)

# 2. Check for Missing Values
print("\n--- 2. Checking Missing Values ---")
print(df.isnull().sum())

# Fill missing values with the mean of the column
df['Hours'] = df['Hours'].fillna(df['Hours'].mean())
print("\nMissing values after mean imputation:")
print(df.isnull().sum())

# 3. Check and Remove Duplicate Rows
print("\n--- 3. Checking Duplicates ---")
print("Duplicate count:", df.duplicated().sum())

df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)

# 4. Save the Clean Dataset
clean_filename = "student_scores_clean.csv"
df.to_csv(clean_filename, index=False)
print(f"\n--- 4. Clean Dataset Saved as '{clean_filename}' ---")

# 5. Final Dataset Summary Statistics
print("\n--- 5. Clean Dataset Statistics ---")
print(df.describe())
```
