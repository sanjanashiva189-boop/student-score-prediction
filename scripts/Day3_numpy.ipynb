```python
# ============================================
# DAY 3: NUMPY TASKS
# ============================================

import numpy as np

# 1. NumPy Arrays Creation
print("--- 1. Creating NumPy Arrays ---")
# 1D Array representing study hours for 5 students
study_hours = np.array([2.5, 5.0, 6.5, 8.0, 3.5])

# 2D Array representing [study_hours, exam_score]
student_data = np.array([
    [2.5, 55],
    [5.0, 78],
    [6.5, 85],
    [8.0, 95],
    [3.5, 62]
])

print("1D Array (Study Hours):", study_hours)
print("\n2D Array (Hours & Scores):\n", student_data)
print("Array Shape:", student_data.shape)


# 2. Indexing and Slicing
print("\n--- 2. Indexing and Slicing ---")
first_student_hours = study_hours[0]
print(f"First student study hours: {first_student_hours}")

# Slice: Extract scores (all rows, 2nd column)
scores = student_data[:, 1]
print("All student scores:", scores)


# 3. Mathematical Operations
print("\n--- 3. Mathematical & Statistical Operations ---")
mean_hours = np.mean(study_hours)
max_score = np.max(scores)
min_score = np.min(scores)
std_dev_scores = np.std(scores)

print(f"Average Study Hours: {mean_hours:.2f} hrs")
print(f"Highest Score: {max_score}")
print(f"Lowest Score: {min_score}")
print(f"Standard Deviation of Scores: {std_dev_scores:.2f}")


# 4. Array-based Calculations (Broadcasting)
print("\n--- 4. Array Calculations ---")
# Suppose each student gets a 5-point bonus score
bonus_scores = scores + 5
print("Scores after +5 bonus points:", bonus_scores)

# Convert study hours to study minutes (Element-wise multiplication)
study_minutes = study_hours * 60
print("Study hours converted to minutes:", study_minutes)
```
