```python
# ============================================
# DAY 11: PREDICTION APP TASKS
# ============================================

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# 1. Load dataset & train model
df = pd.read_csv("data/student_scores_clean.csv")

X = df[['Hours']]
Y = df['Scores']

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, Y_train)

# 2. Interactive Prediction Program
print("=" * 50)
print(" 🎓 STUDENT SCORE PREDICTION APPLICATION 🎓 ")
print("=" * 50)

while True:
    user_input = input("\nEnter study hours (or type 'exit' to quit): ")
    
    if user_input.lower() == 'exit':
        print("\nThank you for using the Student Score Predictor! Goodbye! 👋")
        break
    
    try:
        hours = float(user_input)
        if hours < 0 or hours > 24:
            print("⚠️ Please enter a realistic number of hours between 0 and 24.")
            continue
            
        predicted_score = model.predict([[hours]])[0]
        # Cap prediction at 100 max score
        predicted_score = min(predicted_score, 100.0)
        
        print(f"🎯 Predicted Score for {hours} hours of study: {predicted_score:.2f}%")
        
    except ValueError:
        print("❌ Invalid input! Please enter a valid numerical value.")
```
