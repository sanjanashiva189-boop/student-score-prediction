```python
num1 = 15
num2 = 4

print("--- 2. Arithmetic & Comparison Operators ---")
print(f"Addition ({num1} + {num2}): {num1 + num2}")
print(f"Multiplication ({num1} * {num2}): {num1 * num2}")
print(f"Division ({num1} / {num2}): {num1 / num2}")
print(f"Is {num1} greater than {num2}?: {num1 > num2}\n")
```

```python
print("--- 3. Loops and Conditionals ---")
study_hours_list = [2, 4, 6, 8, 10]

for hours in study_hours_list:
    if hours >= 6:
        status = "Excellent Effort!"
    elif hours >= 4:
        status = "Good Progress"
    else:
        status = "Needs Improvement"
    print(f"Studied {hours} hours -> Status: {status}")
print()
```

```python
def calculate_grade(score):
    """Calculates letter grade based on numerical score."""
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "D"

print("--- 4. Function Demo ---")
sample_score = 82
print(f"Score: {sample_score}% | Grade: {calculate_grade(sample_score)}")
```
