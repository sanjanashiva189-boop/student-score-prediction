```python
# ==============================================================================
# STUDENT SCORE PREDICTION - MASTER PIPELINE
# Codomax Digital Solutions - AI & ML Internship (Day 12)
# ==============================================================================

import os
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

def run_pipeline():
    """
    Executes the end-to-end Machine Learning pipeline:
    1. Automatic Working Directory Adjustment
    2. Data Loading & Inspection
    3. Feature & Target Splitting
    4. Train-Test Data Split
    5. Linear Regression Model Training
    6. Performance Evaluation (MAE, MSE, R² Score)
    """
    
    # -------------------------------------------------------------------------
    # STEP 0: AUTOMATIC DIRECTORY CHECK
    # Fixes FileNotFoundError if running from inside the 'scripts/' folder
    # -------------------------------------------------------------------------
    current_dir = os.getcwd()
    if os.path.basename(current_dir) == 'scripts':
        os.chdir('..')  # Move up to the main project root folder
        print("📁 Adjusted working directory to project root folder.")

    # -------------------------------------------------------------------------
    # STEP 1: LOAD CLEAN DATASET
    # -------------------------------------------------------------------------
    print("\n[1/5] Loading Clean Dataset...")
    
    # Checks for clean dataset first, falls back to original if clean doesn't exist
    data_path = os.path.join("data", "student_scores_clean.csv")
    if not os.path.exists(data_path):
        data_path = os.path.join("data", "student_scores.csv")
        
    df = pd.read_csv(data_path)
    print(f"Loaded File: {data_path}")
    print(f"Dataset Shape: {df.shape} (Rows, Columns)")

    # -------------------------------------------------------------------------
    # STEP 2: DEFINE FEATURES (X) AND TARGET (Y)
    # -------------------------------------------------------------------------
    print("\n[2/5] Extracting Features and Target...")
    
    # Drop missing values if any exist in raw data
    df = df.dropna(subset=['Hours', 'Scores'])
    
    X = df[['Hours']]  # 2D Feature Matrix
    Y = df['Scores']   # 1D Target Vector

    # -------------------------------------------------------------------------
    # STEP 3: TRAIN-TEST SPLIT (80% Train, 20% Test)
    # -------------------------------------------------------------------------
    print("\n[3/5] Splitting Data into Train & Test sets...")
    X_train, X_test, Y_train, Y_test = train_test_split(
        X, Y, test_size=0.2, random_state=42
    )
    print(f"Training set: {len(X_train)} samples | Testing set: {len(X_test)} samples")

    # -------------------------------------------------------------------------
    # STEP 4: MODEL TRAINING
    # -------------------------------------------------------------------------
    print("\n[4/5] Training Linear Regression Model...")
    model = LinearRegression()
    model.fit(X_train, Y_train)
    
    print("✅ Model Training Successful!")
    print(f"Model Intercept (c) : {model.intercept_:.2f}")
    print(f"Model Slope (m)     : {model.coef_[0]:.2f}")
    print(f"Formula: Predicted Score = ({model.coef_[0]:.2f} * Hours) + {model.intercept_:.2f}")

    # -------------------------------------------------------------------------
    # STEP 5: MODEL EVALUATION
    # -------------------------------------------------------------------------
    print("\n[5/5] Evaluating Model Metrics...")
    Y_pred = model.predict(X_test)
    
    mae = mean_absolute_error(Y_test, Y_pred)
    mse = mean_squared_error(Y_test, Y_pred)
    r2 = r2_score(Y_test, Y_pred)

    print("=" * 45)
    print(" 📊 MODEL PERFORMANCE EVALUATION 📊")
    print("=" * 45)
    print(f" Mean Absolute Error (MAE) : {mae:.2f}")
    print(f" Mean Squared Error (MSE)  : {mse:.2f}")
    print(f" Root Mean Squared (RMSE)  : {np.sqrt(mse):.2f}")
    print(f" R² Accuracy Score         : {r2:.4f} ({r2 * 100:.2f}%)")
    print("=" * 45)

# Run the master pipeline
if __name__ == "__main__":
    run_pipeline()
```
